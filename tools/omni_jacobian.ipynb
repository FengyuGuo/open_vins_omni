```python
import sympy as sp
```

```python
x=sp.Symbol('x')
y=sp.Symbol('y')
xi=sp.Symbol('xi')
z_dot=1.0+xi*sp.sqrt(x**2+y**2+1)
u=x/z_dot
v=y/z_dot
r=sp.sqrt(u**2+v**2)
r2=r**2
r4=r2**2
k1=sp.Symbol('k1')
k2=sp.Symbol('k2')
p1=sp.Symbol('p1')
p2=sp.Symbol('p2')
u_d=u*(1+k1*r2+k2*r4)+2*p1*u*v+p2*(r2+x*u**2)
v_d=v*(1+k1*r2+k2*r4)+p1*(r2+x*v**2)+2*p2*u*v
fx=sp.Symbol('fx')
fy=sp.Symbol('fy')
cx=sp.Symbol('cx')
cy=sp.Symbol('cy')
z_u=u_d*fx+cx
z_v=v_d*fy+cy
```
