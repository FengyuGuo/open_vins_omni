```python
import sympy as sp

x=sp.Symbol('x') # normalized x coordinate x = X/Z
y=sp.Symbol('y') # normalized y coordinate y = Y/Z
r=sp.sqrt(x**2+y**2)
r2=r**2
r4=r2**2
r6=r2*r4
k1=sp.Symbol('k1')
k2=sp.Symbol('k2')
p1=sp.Symbol('p1')
p2=sp.Symbol('p2')
k3=sp.Symbol('k3')
u_d=x*(1+k1*r2+k2*r4+k3*r6)+2*p1*x*y+p2*(r2+x*x**2)
v_d=y*(1+k1*r2+k2*r4+k3*r6)+p1*(r2+x*y**2)+2*p2*x*y
fx=sp.Symbol('fx')
fy=sp.Symbol('fy')
cx=sp.Symbol('cx')
cy=sp.Symbol('cy')
z_u=u_d*fx+cx
z_v=v_d*fy+cy
```
